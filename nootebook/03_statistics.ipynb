# Diabetes Risk & Population Health Analytics
## 03 - Statistical Analysis

This notebook performs statistical hypothesis testing to evaluate relationships between diabetes and major clinical and behavioral risk factors.

```python
import pandas as pd
from scipy.stats import mannwhitneyu
from scipy.stats import chi2_contingency
```

## Load Cleaned Dataset

```python
df = pd.read_csv(
    "/drive/diabetes-risk-analytics-project/data/cleaned_diabetes_data.csv"
)
```

## BMI and Diabetes Statistical Test

```python
group1 = df[df["Diabetes_binary"] == 1]["BMI"]
group2 = df[df["Diabetes_binary"] == 0]["BMI"]

stat, p = mannwhitneyu(group1, group2)

print("Mann-Whitney U Statistic:", stat)
print("P-value:", p)
```

### Interpretation

The statistical test indicates a significant difference in BMI distributions between diabetic and non-diabetic individuals, supporting the relationship between obesity and diabetes risk.

## Age and Diabetes Statistical Test

```python
group1 = df[df["Diabetes_binary"] == 1]["Age"]
group2 = df[df["Diabetes_binary"] == 0]["Age"]

stat, p = mannwhitneyu(group1, group2)

print("Mann-Whitney U Statistic:", stat)
print("P-value:", p)
```

### Interpretation

The analysis reveals a statistically significant age difference between diabetic and non-diabetic populations, highlighting age as a critical diabetes risk factor.

## High Blood Pressure and Diabetes Association

```python
contingency_table = pd.crosstab(
    df["HighBP"],
    df["Diabetes_binary"]
)

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-square:", chi2)
print("P-value:", p)
```

### Interpretation

High blood pressure demonstrates a statistically significant association with diabetes prevalence, suggesting overlapping cardiovascular and metabolic risk factors.

## Physical Activity and Diabetes Association

```python
contingency_table = pd.crosstab(
    df["PhysActivity"],
    df["Diabetes_binary"]
)

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-square:", chi2)
print("P-value:", p)
```

### Interpretation

Physical activity levels are significantly associated with diabetes prevalence, reinforcing the importance of lifestyle behavior in population health management.

## Smoking and Diabetes Association

```python
contingency_table = pd.crosstab(
    df["Smoker"],
    df["Diabetes_binary"]
)

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("Chi-square:", chi2)
print("P-value:", p)
```

### Interpretation

Smoking behavior shows a statistically significant relationship with diabetes occurrence, indicating broader lifestyle-related health risks.

## Statistical Analysis Summary

The statistical analysis confirmed several significant relationships between diabetes and key health indicators:

- Higher BMI levels were significantly associated with diabetes.
- Older age groups demonstrated increased diabetes prevalence.
- High blood pressure showed strong association with diabetes occurrence.
- Lifestyle factors such as smoking and physical inactivity were significantly linked to diabetes risk.

These findings support the importance of preventive healthcare strategies and population-level risk management.
