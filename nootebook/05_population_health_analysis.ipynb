# Diabetes Risk & Population Health Analytics
## 05 - Population Health Analysis

This notebook explores population-level health patterns, diabetes prevalence, behavioral risk factors, and cluster-based health segmentation using Pandas.

```python
import pandas as pd
```

## Load Featured Dataset

```python
df = pd.read_csv(
    "/drive/diabetes-risk-analytics-project/data/featured_diabetes_data.csv"
)

df.head()
```

## Diabetes Distribution Analysis

```python
diabetes_dist = (
    df["Diabetes_binary"]
    .value_counts()
    .reset_index()
)

diabetes_dist.columns = [
    "Diabetes_Status",
    "Patient_Count"
]

diabetes_dist["Percentage"] = round(
    diabetes_dist["Patient_Count"]
    / len(df) * 100,
    2
)

diabetes_dist
```

### Insight

Most individuals in the population do not have diabetes. However, a clinically important subset presents diabetic conditions, highlighting opportunities for prevention and chronic disease management.

## Average BMI by Diabetes Status

```python
df.groupby(
    "Diabetes_binary"
)["BMI"].mean().round(2).reset_index()
```

### Insight

Average BMI is considerably higher among diabetic individuals, supporting the well-established relationship between obesity and diabetes risk.

## Population Cluster Distribution

```python
cluster_dist = (
    df["Cluster"]
    .value_counts()
    .reset_index()
)

cluster_dist.columns = [
    "Cluster",
    "Population"
]

cluster_dist
```

### Insight

The clustering model segmented the population into distinct health groups, enabling targeted risk assessment and intervention planning.

## High-Risk Cluster Analysis

```python
cluster_summary = (
    df.groupby("Cluster")
    .agg({
        "BMI":"mean",
        "HighBP":"mean",
        "HighChol":"mean",
        "Diabetes_binary":"mean",
        "GenHlth":"mean"
    })
    .round(2)
)

cluster_summary
```

### Insight

Certain clusters demonstrate elevated diabetes prevalence combined with poorer general health and higher cardiovascular risk indicators.

## Lifestyle Risk Analysis

```python
lifestyle = (
    df.groupby(
        ["PhysActivity","Smoker"]
    )
    .agg(
        Population=("Diabetes_binary","count"),
        Diabetes_Rate=("Diabetes_binary","mean")
    )
    .reset_index()
)

lifestyle["Diabetes_Rate"] = round(
    lifestyle["Diabetes_Rate"] * 100,
    2
)

lifestyle.sort_values(
    "Diabetes_Rate",
    ascending=False
)
```

### Insight

Individuals with unhealthy lifestyle behaviors tend to exhibit higher diabetes prevalence, emphasizing the value of preventive health programs.

## Age-Based Diabetes Risk

```python
age_risk = (
    df.groupby("Age")
    .agg(
        Population=("Diabetes_binary","count"),
        Diabetes_Rate=("Diabetes_binary","mean")
    )
    .reset_index()
)

age_risk["Diabetes_Rate"] = round(
    age_risk["Diabetes_Rate"] * 100,
    2
)

age_risk.sort_values(
    "Diabetes_Rate",
    ascending=False
)
```

### Insight

Older age groups consistently display higher diabetes prevalence, confirming age as one of the strongest population-level risk factors.

## Education and Diabetes

```python
education_analysis = (
    df.groupby("Education")
    .agg(
        Population=("Diabetes_binary","count"),
        Diabetes_Rate=("Diabetes_binary","mean")
    )
    .reset_index()
)

education_analysis["Diabetes_Rate"] = round(
    education_analysis["Diabetes_Rate"] * 100,
    2
)

education_analysis
```

### Insight

Educational attainment appears associated with diabetes prevalence, suggesting that social determinants of health may influence chronic disease outcomes.

## Income and Diabetes

```python
income_analysis = (
    df.groupby("Income")
    .agg(
        Population=("Diabetes_binary","count"),
        Diabetes_Rate=("Diabetes_binary","mean")
    )
    .reset_index()
)

income_analysis["Diabetes_Rate"] = round(
    income_analysis["Diabetes_Rate"] * 100,
    2
)

income_analysis
```

### Insight

Income levels show meaningful variation in diabetes prevalence, highlighting potential socioeconomic disparities in population health.

## Top Risk Profiles

```python
df.groupby(
    ["HighBP","HighChol","PhysActivity"]
)["Diabetes_binary"] \
.mean() \
.sort_values(ascending=False) \
.head(10)
```

### Insight

The highest-risk population profiles combine cardiovascular risk factors and lower levels of healthy lifestyle behaviors.

## Population Health Analysis Summary

Key findings include:

- Diabetes prevalence increases substantially with age.
- Higher BMI levels are strongly associated with diabetes.
- Certain population clusters exhibit elevated clinical risk profiles.
- Lifestyle behaviors such as smoking and physical inactivity are linked to higher diabetes rates.
- Education and income appear related to diabetes prevalence, highlighting the role of social determinants of health.

These findings support population health management, preventive healthcare planning, and risk-based intervention strategies.
