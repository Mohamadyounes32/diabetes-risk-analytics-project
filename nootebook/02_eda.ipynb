This notebook explores:
- Diabetes distribution
- Population health indicators
- Lifestyle and behavioral patterns
- Clinical risk factors
- Relationships between BMI, age, and diabetes risk

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
plt.rcParams["figure.figsize"] = (8,5)
```

```python
main_palette = ["#4C72B0", "#55A868"]
```

## Load Cleaned Dataset

```python
df = pd.read_csv(
    "/drive/diabetes-risk-analytics-project/data/cleaned_diabetes_data.csv"
)

df.head()
```

## Diabetes Distribution

```python
sns.countplot(data=df, x="Diabetes_binary", hue="Diabetes_binary",
              palette=main_palette, legend=False)

plt.show()
```

### Insight

The dataset is highly imbalanced, with non-diabetic individuals representing the majority of the population. This imbalance is important for both statistical interpretation and future machine learning applications.

## BMI Distribution

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="BMI",
    bins=30,
    kde=True
)

plt.title("BMI Distribution")

plt.show()
```

### Insight

BMI values are concentrated between the overweight and obese ranges, suggesting a population with elevated metabolic risk factors associated with diabetes.

## BMI and Diabetes Relationship

```python
sns.boxplot(data=df, x="Diabetes_binary", y="BMI", hue="Diabetes_binary",
            palette=main_palette, legend=False)

plt.show()
```

### Insight

Individuals with diabetes generally exhibit higher BMI values, reinforcing the relationship between obesity and diabetes risk.

## Age and Diabetes Risk

```python
sns.boxplot(
    data=df,
    x="Diabetes_binary",
    y="Age",
    hue="Diabetes_binary",
    palette=main_palette,
    legend=False
)
plt.title("Age by Diabetes Status")
plt.show()
```

### Insight

Diabetes prevalence tends to increase with age, highlighting aging as a major risk factor in population health management.

## High Blood Pressure and Diabetes

```python
sns.countplot(
    data=df,
    x="HighBP",
    hue="Diabetes_binary",
    palette=main_palette
)

plt.title("High Blood Pressure by Diabetes Status")

plt.show()
```

### Insight

Patients with high blood pressure show noticeably higher diabetes prevalence, indicating a strong association between cardiovascular and metabolic conditions.

## Physical Activity and Diabetes

```python
sns.countplot(
    data=df,
    x="PhysActivity",
    hue="Diabetes_binary",
    palette=main_palette
)

plt.title("Physical Activity by Diabetes Status")

plt.show()
```

### Insight

Lower physical activity levels are associated with higher diabetes prevalence, emphasizing the importance of lifestyle interventions in disease prevention.

## Correlation Heatmap

```python
plt.figure(figsize=(14,10))

corr = df.corr()

sns.heatmap(
    corr,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")

plt.show()
```

### Insight

Several health indicators such as BMI, HighBP, and GenHlth demonstrate moderate positive relationships with diabetes, supporting their importance as predictive health factors.

## EDA Summary

Key findings from the exploratory analysis include:

- Diabetes prevalence increases with higher BMI and older age groups.
- High blood pressure is strongly associated with diabetes occurrence.
- Physical inactivity appears linked to elevated diabetes risk.
- Population-level behavioral and health indicators reveal important patterns for public health analysis and preventive care strategies.
