## Load Dataset

```python
import pandas as pd

df = pd.read_csv(
    "/drive/diabetes-risk-analytics-project/data/diabetes_binary_health_indicators_BRFSS2015.csv"
)

df.head()
```

## Initial Data Exploration

```python
df = df.drop_duplicates()
```

```python
df.shape
```

## Convert Binary Columns to Integer Format

```python
binary_cols = [
    "Diabetes_binary",
    "HighBP",
    "HighChol",
    "CholCheck",
    "Smoker",
    "Stroke",
    "HeartDiseaseorAttack",
    "PhysActivity",
    "Fruits",
    "Veggies",
    "HvyAlcoholConsump",
    "AnyHealthcare",
    "NoDocbcCost",
    "DiffWalk",
    "Sex"
]

df[binary_cols] = df[binary_cols].astype(int)
```

## Export Cleaned Dataset

```python
df.to_csv(
    "/drive/diabetes-risk-analytics-project/data/cleaned_diabetes_data.csv",
    index=False
)
```

## Data Cleaning Summary

- No missing values were detected in the dataset.
- Duplicate records were identified and removed.
- Binary indicator columns were converted to integer format for improved analytical consistency.
- The cleaned dataset was exported for downstream analysis and dashboard development.
